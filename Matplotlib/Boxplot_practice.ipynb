Box Plot

```python
'''Q1:Given the marks:
marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 75]
Create a basic box plot with a title "Student Marks".'''
import matplotlib.pyplot as plt
marks = [45, 50, 52, 55, 60, 62, 65, 68, 70, 75]
plt.figure(figsize=(6,4))
plt.boxplot(marks,patch_artist=True)
plt.title("Student Marks")
plt.show()
```

```python
'''Q2:Using:
marks = [35, 40, 42, 45, 50, 52, 55, 60, 65, 70]
Create a horizontal box plot.'''
import matplotlib.pyplot as plt
marks = [35, 40, 42, 45, 50, 52, 55, 60, 65, 70]
plt.figure(figsize=(7,3))
plt.boxplot(marks,patch_artist=True,vert=False,boxprops=dict(color="r",facecolor="skyblue"))
plt.title("Student Marks")
plt.show()
```

```python
'''Q3:Given:
marks = [45, 48, 50, 52, 54, 55, 57, 60, 62, 95]
Create a box plot and determine whether 95 is a possible outlier.'''
import matplotlib.pyplot as plt
marks = [45, 48, 50, 52, 54, 55, 57, 60, 62, 95]
plt.figure(figsize=(5,4))
plt.boxplot(marks,patch_artist=True,boxprops=dict(color="y",facecolor="r"))
plt.title("Student Marks")
plt.show()
```

median=54+55/2=54.5
Q1=50
Q3=60;
IQR=q3-q1
    =60-50=10;    
    Upper fence=Q3+1.5*IQR
        =60+1.5*10=60+15=75;
           
Since 95>75, it is considered as a potential outlier

```python
'''Q4:Create side-by-side box plots for:
class_a = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80]
class_b = [50, 55, 58, 60, 64, 67, 70, 73, 76, 82]
Add labels Class A and Class B.
Then compare their median and spread.'''
import matplotlib.pyplot as plt
class_a = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80]
class_b = [50, 55, 58, 60, 64, 67, 70, 73, 76, 82]
y=[class_a,class_b]
plt.figure(figsize=(5,4))
plt.boxplot(y,patch_artist=True,tick_labels=["Class A","Class B"],
            boxprops=dict(facecolor="r"))

plt.title("Student Marks")
plt.ylabel("Marks")
plt.show()
```

Median of class A is 69

Median of class A is 65.5

Class A has a slightly higher median, while Class B has a slightly wider spread of marks. Neither class shows an obvious outlier.

```python
'''Q5:Given:
class_a = [70, 71, 72, 73, 74, 75, 76, 77, 78]
class_b = [55, 60, 65, 70, 75, 80, 85, 90, 95]
Create a box plot for both classes.
Question: Which class has less variation?'''
import matplotlib.pyplot as plt
class_a = [70, 71, 72, 73, 74, 75, 76, 77, 78]
class_b = [55, 60, 65, 70, 75, 80, 85, 90, 95]
y=[class_a,class_b]
plt.figure(figsize=(5,4))
plt.boxplot(y,patch_artist=True,tick_labels=["Class A","Class B"],
            boxprops=dict(facecolor="#eb8934"))

plt.title("Student Marks")
plt.ylabel("Marks")
plt.show()
```

Class A has less variation because its box and overall range are much smaller than Class B's. Therefore, Class A's marks are more consistent.

```python
'''Q6:Given annual salaries (in thousands):
salary = [25, 28, 30, 32, 35, 36, 38, 40, 42, 45, 80]
Create a box plot and determine:
Whether there is a possible outlier
Whether the distribution appears skewed'''
import matplotlib.pyplot as plt
salary = [25, 28, 30, 32, 35, 36, 38, 40, 42, 45, 80]
plt.figure(figsize=(5,4))
plt.boxplot(salary,patch_artist=True,boxprops=dict(facecolor="#34ebd2"))
plt.title("Salary Distribution")
plt.ylabel("Salary in (k)")
plt.show()
```

Median=36
Q1=30
Q3=42;  
              IQR=42-30=12;
                                     Upper fence=42+1.5*12=42+18=60
           
The salary distribution is right-skewed due to the unusually high salary of 80k, which appears as a potential outlier. The median salary is 36k, and most salaries are concentrated between 25k and 45k.

```python
'''Q9:Given:
data = [10, 12, 13, 15, 16, 18, 20, 21, 22, 24, 40]
Create a box plot and answer:
Is there a possible outlier?
What does the median tell you?
Is the distribution approximately symmetric or skewed?
Which side appears to have the longer tail?'''
import matplotlib.pyplot as plt
data = [10, 12, 13, 15, 16, 18, 20, 21, 22, 24, 40]
plt.figure(figsize=(5,4))
plt.boxplot(data,patch_artist=True,boxprops=dict(facecolor="b"))
plt.title("Data Observation")
plt.show()
```

Median=18                                    
Q1=13                                              
Q3=22                                            
IQR=22-13=9                                                        
Upper fence=22+1.5*9=35.5                                                 
The distribution is right-skewed, with a median of 18. The value 40 is a possible outlier because it exceeds the upper fence of 35.5. The right side has the longer tail.

```python
'''Q10:You have the marks of students from two semesters:
semester_1 = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 75]
semester_2 = [50, 55, 58, 60, 65, 68, 70, 72, 75, 78, 95]
Create a side-by-side box plot and analyze:
Which semester has the higher median?
Which has greater variation?
Are there possible outliers?
Which semester's marks are more consistent?'''
import matplotlib.pyplot as plt
semester_1 = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 75]
semester_2 = [50, 55, 58, 60, 65, 68, 70, 72, 75, 78, 95]
y=[semester_1,semester_2]
plt.figure(figsize=(6,4))
plt.boxplot(y,patch_artist=True,tick_labels=["Semester 1","Semester 2"],
            boxprops=dict(facecolor="#34eb7d"))

plt.title("Student Marks")
plt.ylabel("Marks")
plt.show()
```

Semester 2 has a higher median and greater variation, while Semester 1 has more consistent marks. Neither semester shows any obvious outliers.
