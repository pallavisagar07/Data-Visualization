```python
'''Q1:Create a histogram for these marks:
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90]
Use plt.hist()
Add a title.
Add X and Y labels.'''
import matplotlib.pyplot as plt
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90]
plt.figure(figsize=(6,4))
plt.hist(marks,bins=7,edgecolor="red",color="green")
plt.title("Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
'''Q2:Create a histogram using:
ages = [18, 19, 20, 21, 22, 20, 19, 23, 24, 21, 22, 20, 25, 26, 21, 19]
Use bins=5.
Add a grid.
Give the plot an appropriate title.'''
import matplotlib.pyplot as plt
ages = [18, 19, 20, 21, 22, 20, 19, 23, 24, 21, 22, 20, 25, 26, 21, 19]
plt.figure(figsize=(6,4))
plt.hist(ages,bins=5,color="blue")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of People")
plt.grid(True,alpha=0.6)
plt.show()
```

```python
'''Q6:You have marks from two classes:
class_a = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80]
class_b = [50, 55, 58, 60, 64, 67, 70, 73, 76, 82]

Create two histograms on the same graph.
Requirements:
Use different colors.
Use alpha so both distributions are visible.
Add a legend.
Add a title.'''
import matplotlib.pyplot as plt
class_a = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80]
class_b = [50, 55, 58, 60, 64, 67, 70, 73, 76, 82]
plt.figure(figsize=(6,4))
plt.hist(class_a,color="#a83266",label="Class A",alpha=0.7)
plt.hist(class_b,color="#4832a8",label="Class B",alpha=0.7)
plt.title("Two Class Comparison")
plt.xlabel("Marks")
plt.ylabel("No. of Students")
plt.legend()
plt.show()
```

```python
'''Q8:Create a histogram for:
data = [10, 12, 13, 14, 15, 15, 16, 17, 18, 18,
        19, 20, 21, 22, 23, 25, 30, 35]

Then answer:
Where are most values concentrated?
Is the distribution approximately symmetric or skewed?
Are there any unusually high values?
This one is about reading the histogram, not just creating it.'''
import matplotlib.pyplot as plt
data = [10, 12, 13, 14, 15, 15, 16, 17, 18, 18,
        19, 20, 21, 22, 23, 25, 30, 35]
plt.figure(figsize=(6,4))
plt.hist(data,color="#a8325a",edgecolor="black")
plt.title("Distribution Analysis")
plt.xlabel("Data")
plt.ylabel("Number of Observations")
plt.show()
```

Most observations are around 10-23.
The Distribution is right-skewed.
The unusually high values are around 31-35.
