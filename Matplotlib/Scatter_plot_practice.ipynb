```python
'''Q1:Create a scatter plot showing:
x = Study hours: [1, 2, 3, 4, 5, 6]
y = Marks: [45, 50, 55, 65, 70, 80]
Add title and axis labels.'''
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5, 6]
y = [45, 50, 55, 65, 70, 80]
sizes=[40,40,40,40,40,40]
plt.figure(figsize=(6,4))
plt.scatter(x,y,color="red",marker="o",s=sizes)
plt.title("Study Hours Vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
'''Q2:Plot:
Temperature: [20, 22, 25, 27, 30, 32, 35]
Sales: [50, 55, 65, 72, 85, 95, 110]
Use a different marker such as ^.'''
import matplotlib.pyplot as plt
Temperature= [20, 22, 25, 27, 30, 32, 35]
Sales= [50, 55, 65, 72, 85, 95, 110]
plt.figure(figsize=(6,4))
plt.scatter(Temperature,Sales,marker="^")
plt.title("Temperature vs Ice Cream Sales")
plt.xlabel("Temperature")
plt.ylabel("Sales")
plt.show()
```

```python
'''Q3:Create a scatter plot using:
Height: [150, 155, 160, 165, 170, 175, 180]
Weight: [45, 48, 52, 55, 60, 65, 70]
Add a grid.'''
import matplotlib.pyplot as plt
Height= [150, 155, 160, 165, 170, 175, 180]
Weight= [45, 48, 52, 55, 60, 65, 70]
plt.figure(figsize=(6,4))
plt.scatter(Height,Weight,color="purple")
plt.title("Height Vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.grid(True,alpha=0.5)
plt.show()
```

```python
'''Q4:Plot:
Advertising spend: [10, 20, 30, 40, 50, 60]
Sales: [100, 130, 150, 180, 210, 250]
Change the marker size using s.'''
import matplotlib.pyplot as plt
x=[10, 20, 30, 40, 50, 60]
y= [100, 130, 150, 180, 210, 250]
sizes=[110,120,140,130,150,160]
plt.figure(figsize=(6,4))
plt.scatter(x,y,s=sizes,marker="*",color="green")
plt.title("Advertising Vs Sales")
plt.xlabel("Advertising Spend")
plt.ylabel("Sales")
plt.show()
```

```python
'''Q7:Plot two groups of students:

Group A: Hours [2, 3, 4, 5, 6], Marks [50, 55, 65, 70, 80]
Group B: Hours [2, 3, 4, 5, 6], Marks [45, 52, 60, 68, 75]
Give the two groups different markers and add a legend.'''
import matplotlib.pyplot as plt
x=[2, 3, 4, 5, 6]
y1=[50, 55, 65, 70, 80]
y2=[45, 52, 60, 68, 75]
plt.figure(figsize=(6,4))
plt.scatter(x,y1,color="blue",label="Group 1 Marks")
plt.scatter(x,y2,color="red",label="Group 2 Marks")
plt.title("Hours Vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

```python
'''Q9:Given:
Price: [10, 15, 20, 25, 30, 35, 40, 45]
Sales: [500, 450, 400, 350, 300, 250, 180, 120]
Create a scatter plot.
Experiment with marker size, transparency, and marker style.
Add a title explaining the relationship.'''
import matplotlib.pyplot as plt
x=[10, 15, 20, 25, 30, 35, 40, 45]
y=[500, 450, 400, 350, 300, 250, 180, 120]
plt.figure(figsize=(7,4))
sizes=[210,230,220,240,250,210,220,230]
plt.scatter(x,y,color="#eb3443",marker="*",s=sizes,alpha=0.6)
plt.title("Product Price vs Sales")
plt.xlabel("Price")
plt.ylabel("Sales")
plt.show()
```

```python
'''Q10:Create your own dataset containing 10 students with:
Study hours
Marks
Attendance percentage

Create a scatter plot of Study Hours vs Marks where:
Marker size represents attendance.
Add title, labels, grid, and legend.
Try to visually determine whether more study hours are associated with higher marks.'''
import matplotlib.pyplot as plt
Study_Hours= [1, 2, 3, 4, 5, 6]
Marks= [45, 50, 55, 65, 70, 80]
Attendance=[75,78,85,93,95,86]
plt.figure(figsize=(7,4))
plt.scatter(Study_Hours,Marks,color="#eb3443",marker="*",s=Attendance,label="Students")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.legend()
plt.show()
```
