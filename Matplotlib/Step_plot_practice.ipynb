```python
'''Q1:Create a step plot for:
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
Add a title, x-label, and y-label.'''
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.figure(figsize=(6,4))
plt.step(x,y)
plt.title("Basic Step Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
'''Q2:where Parameter
Using the same data, create three step plots using:
where='pre'
where='post'
where='mid'
Understand how the position of the steps changes.'''
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.step(x, y, where='pre',marker="o")
plt.title("Step Plot - Pre")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()

plt.figure(figsize=(8,4))
plt.step(x, y, where='post',marker="o")
plt.title("Step Plot - Post")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()

plt.figure(figsize=(8,4))
plt.step(x, y, where='mid',marker="o")
plt.title("Step Plot - Mid")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()
```

```python
'''Q3:Create a step plot showing weekly sales:
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [20, 35, 30, 45, 40, 60, 55]
Add markers and a grid.'''
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [20, 35, 30, 45, 40, 60, 55]

plt.step(days,sales,marker="o",color="red")
plt.title("Sales Tracking")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
'''Q4:Plot temperature throughout the day:
hours = [6, 9, 12, 15, 18, 21]
temperature = [18, 22, 28, 31, 27, 23]
Use where='mid'.'''
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
hours = [6, 9, 12, 15, 18, 21]
temperature = [18, 22, 28, 31, 27, 23]
plt.step(hours,temperature,where='mid',marker="o",color="maroon")
plt.title("Temperature tracker")
plt.xlabel("Hours")
plt.ylabel("Temperature")
plt.grid()
plt.show()
```

```python
'''Q5:Create a step plot showing the number of active users:
hours = [9, 10, 11, 12, 13, 14, 15]
users = [100, 150, 200, 180, 250, 300, 280]
Customize:line width, marker, linestyle, title,grid'''
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
hours = [9, 10, 11, 12, 13, 14, 15]
users = [100, 150, 200, 180, 250, 300, 280]
plt.step(hours,users,marker="o",color="#14c4ad",linewidth=2,linestyle=":")
plt.title("Internet Users")
plt.xlabel("Hours")
plt.ylabel("Users")
plt.grid(alpha=0.5)
plt.show()
```

```python
'''Q6:Compare two products:
months = [1, 2, 3, 4, 5, 6]
product_A = [10, 15, 20, 25, 30, 35]
product_B = [12, 12, 18, 22, 28, 32]
Plot both using plt.step() and add a legend.'''
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
months = [1, 2, 3, 4, 5, 6]
product_A = [10, 15, 20, 25, 30, 35]
product_B = [12, 12, 18, 22, 28, 32]
plt.step(months,product_A,marker="o",color="#14c4ad",linewidth=2,linestyle=":",label="Product A")
plt.step(months,product_B,marker="*",color="#c41440",linewidth=2,label="Product B",ms=10)

plt.title("Product A Vs Product B")
plt.xlabel("Months")
plt.ylabel("Products")
plt.legend()
plt.grid(alpha=0.5)
plt.show()
```

```python
'''Q7:Create a step plot for inventory:
days = [1, 2, 3, 4, 5, 6, 7]
stock = [100, 100, 80, 80, 60, 60, 40]
Use where='post' and clearly label when stock decreases.'''
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
days = [1, 2, 3, 4, 5, 6, 7]
stock = [100, 100, 80, 80, 60, 60, 40]
plt.step(days,stock,where='post',marker="o",color="#141ac4")
plt.annotate("Stock decreased", (3, 80), xytext=(3.5, 90),
             arrowprops=dict(arrowstyle="->"))
plt.title("Inventory Levels")
plt.xlabel("Days")
plt.ylabel("Stocks")
plt.grid(linestyle="--",alpha=0.6)
plt.show()
```

```python
'''Q8: Step Plot + Normal Plot
Plot the following data twice:
x = [0, 1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6, 9]
Create:
one normal line plot
one step plot
Put them in a 1×2 subplot dashboard and compare them.'''
import matplotlib.pyplot as plt
x = [0, 1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 6, 9]

fix,ax=plt.subplots(1,2,figsize=(9,4))
ax[0].plot(x,y,marker="o",color="#c41469")
ax[0].set_title("Line Plot")
ax[0].set_xlabel("X")
ax[0].set_ylabel("Y")
ax[0].grid(linestyle="--",alpha=0.6)

ax[1].step(x,y,marker="o",color="#141ac4")
ax[1].set_title("Step Plot")
ax[1].set_xlabel("X")
ax[1].set_ylabel("Y")
ax[1].grid(linestyle="--",alpha=0.6)

plt.show()
```

```python
'''Q9:Create a step plot:
hours = [8, 10, 12, 14, 16, 18, 20]
visitors = [50, 80, 120, 200, 150, 100, 70]
Add a horizontal line at 100 visitors using axhline().
Make the chart visually indicate which periodsare above the threshold.'''
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
hours = [8, 10, 12, 14, 16, 18, 20]
visitors = [50, 80, 120, 200, 150, 100, 70]
plt.step(hours,visitors,marker="o",color="#141ac4")
plt.axhline(100, linestyle="--", label="Threshold: 100 Visitors")
plt.title("Website Traffic Threshold")
plt.xlabel("Hours")
plt.ylabel("Visitors")
plt.legend()
plt.grid(linestyle="--",alpha=0.6)
plt.show()
```

```python
'''Q10:A SaaS product changes its monthly price at different points:
months = [1, 3, 6, 9, 12]
price = [199, 299, 399, 499, 599]
Create a professional step plot that:
uses where='post'
shows markers
displays the price changes
has proper title and axis labels
includes a grid
highlights the highest price
uses annotations to show the price at each stage'''
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
months = [1, 3, 6, 9, 12]
price = [199, 299, 399, 499, 599]
plt.step(months,price,where='post',marker="o",color="#bd092a")
plt.scatter(12, 599, s=100)
for x, y in zip(months, price):
    plt.annotate(f"₹{y}", (x, y), xytext=(0, 10),
                 textcoords="offset points", ha="center")
plt.title("Subscription Pricing")
plt.xlabel("Months")
plt.ylabel("Price")
plt.grid(linestyle="--",alpha=0.6)
plt.show()
```
