```python
'''Q9:Percentage Stack Plot
Create a stack plot for:
Python: 40, 45, 50, 55, 60
Java: 30, 28, 25, 22, 20
C++: 20, 18, 15, 13, 12
JavaScript: 10, 9, 10, 10, 8
Convert the values into percentages first.
Create a 100% stacked area plot where the total height is always 100%.'''
import matplotlib.pyplot as plt
months = [1, 2, 3, 4, 5]
Python=[ 40, 45, 50, 55, 60]
Java=[30, 28, 25, 22, 20]
C_plus_plus=[ 20, 18, 15, 13, 12]
JavaScript=[10, 9, 10, 10, 8]
l=["Python","Java","C++","Javascript"]
plt.figure(figsize=(6,4))
plt.stackplot(months,Python,Java, C_plus_plus,JavaScript,labels=l,colors=["#64ede6","#ed072d","#0e9609","#cf1396"])
plt.title("Percentage Stack Plot")
plt.xlabel("Languages")
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)
plt.legend(loc=4)
plt.show()
```
