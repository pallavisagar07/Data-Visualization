```python
'''Q1:Use the following data:
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 180, 160, 200]

products = ["Laptop", "Phone", "Tablet", "Watch"]
product_sales = [45, 70, 35, 25]

languages = ["Python", "Java", "C++", "JavaScript"]
popularity = [40, 25, 20, 15]

study_hours = [2, 3, 4, 5, 6, 7, 8, 9]
marks = [45, 50, 55, 62, 68, 72, 80, 88]
Create a 2×2 subplot containing:

Line chart → Monthly sales
Bar chart → Product sales
Pie chart → Language popularity
Scatter plot → Study hours vs marks
Give every subplot a title.'''
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 180, 160, 200]

products = ["Laptop", "Phone", "Tablet", "Watch"]
product_sales = [45, 70, 35, 25]

languages = ["Python", "Java", "C++", "JavaScript"]
popularity = [40, 25, 20, 15]

study_hours = [2, 3, 4, 5, 6, 7, 8, 9]
marks = [45, 50, 55, 62, 68, 72, 80, 88]


# Create 2 rows and 2 columns
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# 1. Line Plot
ax[0, 0].plot(months, sales, marker="o")
ax[0, 0].set_title("Monthly Sales")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales")
ax[0, 0].grid(True)

# 2. Bar Plot
ax[0, 1].bar(products, product_sales)
ax[0, 1].set_title("Product Sales")
ax[0, 1].set_xlabel("Product")
ax[0, 1].set_ylabel("Sales")

# 3. Pie Chart
ax[1, 0].pie(popularity, labels=languages, autopct="%1.1f%%")
ax[1, 0].set_title("Programming Language Popularity")

# 4. Scatter Plot
ax[1, 1].scatter(study_hours, marks)
ax[1, 1].set_title("Study Hours vs Marks")
ax[1, 1].set_xlabel("Study Hours")
ax[1, 1].set_ylabel("Marks")
ax[1, 1].grid(True)

# Overall title
fig.suptitle("Data Visualization Dashboard", fontsize=16)

plt.tight_layout()
plt.show()




```

```python
'''Q2:Compare Two Classes
class_a = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80,
           82, 85, 88, 90, 92]

class_b = [48, 52, 55, 58, 60, 63, 65, 67, 70, 73,
           75, 78, 80, 82, 85]

Question: Create 2 subplots side by side:

Left → Histogram of Class A marks
Right → Histogram of Class B marks

Add titles, x-axis labels, y-axis labels, and grids.'''
import matplotlib.pyplot as plt

class_a = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80,
           82, 85, 88, 90, 92]

class_b = [48, 52, 55, 58, 60, 63, 65, 67, 70, 73,
           75, 78, 80, 82, 85]

# Create 1 row and 2 columns
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# First subplot - Class A
ax[0].hist(class_a, bins=5, color="skyblue", edgecolor="black")
ax[0].set_title("Class A Marks")
ax[0].set_xlabel("Marks")
ax[0].set_ylabel("Number of Students")
ax[0].grid(axis="y", alpha=0.3)

# Second subplot - Class B
ax[1].hist(class_b, bins=5, color="orange", edgecolor="black")
ax[1].set_title("Class B Marks")
ax[1].set_xlabel("Marks")
ax[1].set_ylabel("Number of Students")
ax[1].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()


```

```python
'''Q4:Create a 2×2 subplot grid containing:
Line plot → Study hours
Bar plot → Marks
Scatter plot → Study hours vs Marks
Pie chart → Subject-wise marks

Data:

subjects = ["Python", "Maths", "SQL", "Statistics"]
study_hours = [2, 4, 3, 5]
marks = [65, 80, 72, 88]'''
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,2, figsize=(10,7))
subjects = ["Python", "Maths", "SQL", "Statistics"]
study_hours = [2, 4, 3, 5]
marks = [65, 80, 72, 88]

#First subplot-Line Chart
ax[0,0].plot(subjects, study_hours, marker="o",color="#f542f2")
ax[0,0].set_title("Study Hours by Subject")
ax[0,0].set_xlabel("Subjects")
ax[0,0].set_ylabel("Study hours")
ax[0,0].grid(True, linestyle="--",alpha=0.7)

#Second subplot-Bar Chart
ax[0,1].bar(subjects,marks,color=["#4287f5","#f542ec","#f5e342","#f54260"],edgecolor="black")
ax[0,1].set_title("Marks by Subject")
ax[0,1].set_xlabel("Subjects")
ax[0,1].set_ylabel("Marks")
ax[0,1].grid(True, linestyle="--",alpha=0.6)

#Third subplot-Scatter Plot
ax[1,0].scatter(study_hours,marks,marker="o",color="#e80c30")
ax[1,0].set_title("Study hours Vs Marks")
ax[1,0].set_xlabel("Study Hours")
ax[1,0].set_ylabel("Marks")
ax[1,0].grid(True, linestyle="--",alpha=0.7)

#Fourth subplot-Pie Chart
ax[1,1].pie(marks,labels=subjects,autopct="%1.0f%%",colors=["#e01d3e","#e01dc6","#05a3a6","#869e0b"])
ax[1,1].set_title("Subject Wise Marks")

plt.tight_layout()
plt.show()
```

```python
'''Q5:Sales Dashboard

Create 4 subplots (2×2) for a small sales dashboard:

Line chart → Monthly revenue
Bar chart → Monthly expenses
Scatter plot → Revenue vs Expenses
Pie chart → Revenue distribution

Data:

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [50, 65, 70, 80, 95, 110]
expenses = [30, 35, 40, 42, 50, 55]'''

import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,2, figsize=(10,7))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [50, 65, 70, 80, 95, 110]
expenses = [30, 35, 40, 42, 50, 55]

#First subplot-Line Chart
ax[0,0].plot(months,revenue, marker="o",color="#0b7e9e")
ax[0,0].set_title("Monthly Revenue")
ax[0,0].set_xlabel("Months")
ax[0,0].set_ylabel("Revenue")
ax[0,0].grid(True, linestyle="dotted",alpha=0.7)

#Second subplot-Bar Chart
ax[0,1].bar(months,expenses,color=["#0785a8","#a0a807","#2a0785","#85071c"])
ax[0,1].set_title("Monthly Expenses")
ax[0,1].set_xlabel("Months")
ax[0,1].set_ylabel("Expenses")
ax[0,1].grid(axis="y",alpha=0.6)

#Third subplot-Scatter Plot
ax[1,0].scatter(revenue,expenses,marker="o",color="#c90e2d",s=100)
ax[1,0].set_title("Revenue vs Expenses")
ax[1,0].set_xlabel("Revenue")
ax[1,0].set_ylabel("Expenses")
ax[1,0].grid(True, linestyle="--",alpha=0.7)

#Fourth subplot-Pie Chart
ax[1,1].pie(revenue,labels=months,autopct="%1.0f%%",colors=["#e01d3e","#e01dc6","#05a3a6","#869e0b"])
ax[1,1].set_title("Revenue distribution")

fig.suptitle("Sales Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
'''Q7:Create 2×2 subplots to compare Product A and Product B.

Line plot → Product A sales
Line plot → Product B sales
Bar chart → Compare both products
Scatter plot → Product A vs Product B

Data:
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [40, 45, 50, 55, 60, 65]
product_B = [35, 42, 48, 52, 58, 62]'''
import matplotlib.pyplot as plt
import numpy as np
fig,ax=plt.subplots(2,2, figsize=(10,7))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [40, 45, 50, 55, 60, 65]
product_B = [35, 42, 48, 52, 58, 62]


#First subplot-Line Chart
ax[0,0].plot(months, product_A, marker="o", label="Product A")
ax[0,0].set_title("Product A Monthly Sales")
ax[0,0].set_xlabel("Months")
ax[0,0].set_ylabel("Sales")
ax[0,0].legend()
ax[0,0].grid(alpha=0.5)

#Second subplot- Line Chart
ax[0,1].plot(months,product_B,color="#85071c",marker="o",label="Product B")
ax[0,1].set_title("Monthly Product B Sales")
ax[0,1].set_xlabel("Months")
ax[0,1].set_ylabel("Product B Sales")
ax[0,1].legend()
ax[0,1].grid(alpha=0.6)

#Third subplot-Bar Chart
x = range(len(months))
width = 0.35

ax[1,0].bar([i - width/2 for i in x], product_A,
            width=width, label="Product A")

ax[1,0].bar([i + width/2 for i in x], product_B,
            width=width, label="Product B")

ax[1,0].set_title("Product A vs Product B Sales")
ax[1,0].set_xlabel("Months")
ax[1,0].set_ylabel("Sales")
ax[1,0].set_xticks(x)
ax[1,0].set_xticklabels(months)
ax[1,0].legend()

#Fourth subplot-Scatter Plot
ax[1,1].scatter(product_A,product_B,color="orange")
ax[1,1].set_title("Product A Vs Product B")
ax[1,1].set_xlabel("Product A")
ax[1,1].set_ylabel("Product B")
ax[1,1].grid()

fig.suptitle("Product Sales Comparison", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
'''Q8:Create a 2×2 subplot layout where the first subplot occupies two columns.

Use:
Large subplot → Line chart of website visitors
Small subplot → Bar chart of signups
Small subplot → Scatter plot of visitors vs signups

Data:
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [500, 650, 700, 800, 950, 1100, 1250]
signups = [20, 28, 32, 40, 48, 55, 65]
Hint: Practice using subplot2grid() or GridSpec.'''
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

visitors = [500, 650, 700, 800, 950, 1100, 1250]
signups = [20, 28, 32, 40, 48, 55, 65]

# Create figure
fig = plt.figure(figsize=(10, 7))

# Create 2×2 grid
gs = fig.add_gridspec(2, 2)

# 1. Large subplot — Line Chart
ax1 = fig.add_subplot(gs[0, :])

ax1.plot(days, visitors, marker="o")
ax1.set_title("Weekly Website Visitors")
ax1.set_xlabel("Days")
ax1.set_ylabel("Visitors")
ax1.grid(alpha=0.5)

# 2. Small subplot — Bar Chart
ax2 = fig.add_subplot(gs[1, 0])

ax2.bar(days, signups)
ax2.set_title("Daily Signups")
ax2.set_xlabel("Days")
ax2.set_ylabel("Signups")

# 3. Small subplot — Scatter Plot
ax3 = fig.add_subplot(gs[1, 1])

ax3.scatter(visitors, signups)
ax3.set_title("Visitors vs Signups")
ax3.set_xlabel("Visitors")
ax3.set_ylabel("Signups")
ax3.grid(alpha=0.5)

# Overall title
fig.suptitle("Website Performance Dashboard", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
'''Q9:Create a 2×2 dashboard using the following data:

Line plot → Revenue trend
Bar plot → Profit
Scatter plot → Revenue vs Profit
Histogram → Revenue distribution

Also practice:figsize,suptitle(),set_title(),set_xlabel(),set_ylabel(),legend(),grid(),
tight_layout()

Data:
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
revenue = [120, 135, 150, 145, 170, 185,
           200, 210, 195, 220, 240, 260]
profit = [30, 35, 40, 38, 45, 50,
          55, 60, 52, 65, 70, 80]'''
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue = [120, 135, 150, 145, 170, 185,
           200, 210, 195, 220, 240, 260]

profit = [30, 35, 40, 38, 45, 50,
          55, 60, 52, 65, 70, 80]

# Create 2 × 2 subplots
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# 1. Line Plot - Revenue Trend
ax[0, 0].plot(months, revenue, marker="o",color="green")
ax[0, 0].set_title("Monthly Revenue Trend")
ax[0, 0].set_xlabel("Months")
ax[0, 0].set_ylabel("Revenue")
ax[0, 0].grid(alpha=0.5)

# 2. Bar Plot - Profit
ax[0, 1].bar(months, profit,color="maroon")
ax[0, 1].set_title("Monthly Profit")
ax[0, 1].set_xlabel("Months")
ax[0, 1].set_ylabel("Profit")

# 3. Scatter Plot - Revenue vs Profit
ax[1, 0].scatter(revenue, profit)
ax[1, 0].set_title("Revenue vs Profit")
ax[1, 0].set_xlabel("Revenue")
ax[1, 0].set_ylabel("Profit")
ax[1, 0].grid(alpha=0.5)

# 4. Histogram - Revenue Distribution
ax[1, 1].hist(revenue, bins=5, edgecolor="black",color="y")
ax[1, 1].set_title("Revenue Distribution")
ax[1, 1].set_xlabel("Revenue")
ax[1, 1].set_ylabel("Frequency")

# Overall title
fig.suptitle("Business Performance Dashboard", fontsize=16)

# Adjust spacing
plt.tight_layout()

plt.show()
```

```python
'''Q10:Monthly Business Analysis Dashboard

Use the following data:
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 140, 180, 200, 220]
expenses = [80, 90, 85, 100, 110, 120]
customers = [50, 65, 60, 75, 85, 95]

Create a 2×2 subplot dashboard:
Line Plot → Monthly Sales
Bar Plot → Monthly Expenses
Scatter Plot → Customers vs Sales
Line Plot → Customers over Months

Also add:
Appropriate titles
X and Y labels
marker where suitable
grid() where suitable
fig.suptitle("Monthly Business Analysis")
plt.tight_layout()'''
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [120, 150, 140, 180, 200, 220]
expenses = [80, 90, 85, 100, 110, 120]
customers = [50, 65, 60, 75, 85, 95]

# Create 2 × 2 subplots
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# 1. Line Plot - Monthly Sales
ax[0, 0].plot(months, sales, marker="o")
ax[0, 0].set_title("Monthly Sales")
ax[0, 0].set_xlabel("Months")
ax[0, 0].set_ylabel("Sales")
ax[0, 0].grid(alpha=0.5)

# 2. Bar Plot - Monthly Expenses
ax[0, 1].bar(months, expenses, edgecolor="black")
ax[0, 1].set_title("Monthly Expenses")
ax[0, 1].set_xlabel("Months")
ax[0, 1].set_ylabel("Expenses")

# 3. Scatter Plot - Customers vs Sales
ax[1, 0].scatter(customers, sales)
ax[1, 0].set_title("Customers vs Sales")
ax[1, 0].set_xlabel("Customers")
ax[1, 0].set_ylabel("Sales")
ax[1, 0].grid(alpha=0.5)

# 4. Line Plot - Customers over Months
ax[1, 1].plot(months, customers, marker="o")
ax[1, 1].set_title("Monthly Customers")
ax[1, 1].set_xlabel("Months")
ax[1, 1].set_ylabel("Customers")
ax[1, 1].grid(alpha=0.5)

# Overall title
fig.suptitle("Monthly Business Analysis", fontsize=16)

# Adjust spacing
plt.tight_layout()

plt.show()
```
